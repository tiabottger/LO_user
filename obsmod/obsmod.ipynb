```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import pickle
from pathlib import Path
import xarray as xr
from lo_tools import Lfun
from lo_tools import plotting_functions as pfun

Ldir = Lfun.Lstart()
```

```python
pkl_path = (
    Path(Ldir["LOo"])
    / "obsmod"
    / "combined_bottle_2014_cas7_t1_x11ab_ssc_ssmpnnl_ecology.pkl"
)

with open(pkl_path, "rb") as f:
    data = pickle.load(f)

ecology = data["ecology"]
ssm = data["ssm"]
obs = data["obs"]
```

```python
# load ecology matched observation dataset

ecology_obs_match = pd.read_excel(Path(Ldir["LOu"]) / "intermodel_comparison" / "SSM_Opt2_Year_2014_Exist_Model_Obs_Paired.xlsx")
```

```python
ecology_obs_match.head()
```

```python
# number of unique stations in the ecology matched observation dataset
num_unique_stations = ecology_obs_match["Station"].nunique()
print(f"Number of unique stations in the ecology matched observation dataset: {num_unique_stations}")

# number of unique stations in our observational dataset

stations = (
    obs[['name', 'lon', 'lat', 'source']]
    .drop_duplicates()
)

num_unique_stations_obs = len(stations)
print(f"Number of unique stations in our obs dataset: {num_unique_stations_obs}")
```

```python
mean_vert_diff_ecology = ecology["vertical_difference_m"].mean()
mean_vert_diff_ssmpnnl = ssm["vertical_difference_m"].mean()

print(f"Mean vertical difference for Ecology: {mean_vert_diff_ecology:.2f} m")
print(f"Mean vertical difference for SSM PNNL: {mean_vert_diff_ssmpnnl:.2f} m")
```

```python
mean_horiz_dist_ecology = ecology["horizontal_distance_km"].mean()
mean_horiz_dist_ssmpnnl = ssm["horizontal_distance_km"].mean()

print(f"Mean horizontal distance for Ecology: {mean_horiz_dist_ecology:.2f} km")
print(f"Mean horizontal distance for SSM PNNL: {mean_horiz_dist_ssmpnnl:.2f} km")
```

```python
mean_time_diff_ecology = ecology["time_difference_min"].mean()
mean_time_diff_ssmpnnl = ssm["time_difference_min"].mean()

print(f"Mean time difference for Ecology: {mean_time_diff_ecology:.2f} min")
print(f"Mean time difference for SSM PNNL: {mean_time_diff_ssmpnnl:.2f} min")
```
